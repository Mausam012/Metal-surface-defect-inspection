# Dataset exploration
Counts images per class and inspects intensity statistics. Point `ROOT` at your own authorized dataset. Nothing here is a model result.

```python
from pathlib import Path
import cv2, numpy as np
ROOT = Path('../dataset')            # dataset/{train,validation,test}/{normal,defective}
EXT = {'.png','.jpg','.jpeg','.tif','.tiff'}
for split in ('train','validation','test'):
    for cls in ('normal','defective'):
        n = sum(1 for p in (ROOT/split/cls).glob('*') if p.suffix.lower() in EXT)
        print(f'{split:<11}{cls:<10}{n}')
```

## Image statistics
Bit depth, size and intensity range for the first few images of a folder.

```python
folder = ROOT/'train'/'normal'
for p in sorted(folder.glob('*'))[:5]:
    img = cv2.imread(str(p), cv2.IMREAD_UNCHANGED)
    if img is None: continue
    print(p.name, img.shape, img.dtype, int(img.min()), int(img.max()), round(float(img.mean()),1))
```

## Leakage checklist
- Do images from the same specimen / weld series appear in more than one split?
- Were crops or augmented copies made *before* splitting?
- Is the class balance similar in validation and test?
